```python
import arcpy
arcpy.env.overwriteOutput = True
```

```python
folder_path = r"C:\Users\arjav\DevSource\sgsup-arjav-222\lab\lab05\Lab05_Data\Output"
gdb_name = "Lab05.gdb"
gdb_path = folder_path+"/"+gdb_name
arcpy.management.CreateFileGDB(folder_path, gdb_name)
```

```python
csv_file = "C:/Users/arjav/DevSource/sgsup-arjav-222/lab/lab05/Lab05_Data/Input/garages.csv"
```

```python
arcpy.management.MakeXYEventLayer(
    table=csv_file,
    in_x_field="X",
    in_y_field="Y",
    out_layer="GaragePoints"
)

arcpy.conversion.FeatureClassToGeodatabase(
    Input_Features="GaragePoints",
    Output_Geodatabase=r"C:\Users\arjav\DevSource\sgsup-arjav-222\lab\lab05\Lab05_Data\Output\Lab05.gdb"
)
```

```python
arcpy.management.CopyFeatures(
    in_features=r"C:\Users\arjav\DevSource\sgsup-arjav-222\lab\lab05\Lab05_Data\Input\ASU_Tempe_Campus_Bldgs.shp",
    out_feature_class=r"C:\Users\arjav\DevSource\sgsup-arjav-222\lab\lab05\Lab05_Data\Output\Lab05.gdb/BuildingsLayer")
```

```python
building_layer = r"C:\Users\arjav\DevSource\sgsup-arjav-222\lab\lab05\Lab05_Data\Output\Lab05.gdb/BuildingsLayer"
GaragePoints = r"C:\Users\arjav\DevSource\sgsup-arjav-222\lab\lab05\Lab05_Data\Output\Lab05.gdb/GaragePoints"

arcpy.Describe(building_layer).spatialReference
```

```python
arcpy.management.Project(
    in_dataset=GaragePoints,
    out_dataset=r"C:\Users\arjav\DevSource\sgsup-arjav-222\lab\lab05\Lab05_Data\Output\Lab05.gdb/GaragePoints_Projected",
    out_coor_system=arcpy.Describe(building_layer).spatialReference
)
```

```python
garageBuffer = arcpy.analysis.Buffer(
    in_features=r"C:\Users\arjav\DevSource\sgsup-arjav-222\lab\lab05\Lab05_Data\Output\Lab05.gdb/GaragePoints_Projected",
    out_feature_class=r"C:\Users\arjav\DevSource\sgsup-arjav-222\lab\lab05\Lab05_Data\Output\Lab05.gdb/BuildingBuffers",
    buffer_distance_or_field=500,
)
```

```python
intersection = arcpy.analysis.Intersect(
    in_features=[garageBuffer,building_layer],
    out_feature_class=r"C:\Users\arjav\DevSource\sgsup-arjav-222\lab\lab05\Lab05_Data\Output\Lab05.gdb/IntersectedFeatures",
)
```

```python
arcpy.conversion.TableToTable(
    in_rows=intersection,
    out_path=r"C:\Users\arjav\DevSource\sgsup-arjav-222\lab\lab05\Lab05_Data\Output\Lab05.gdb",
    out_name="nearbyBuildings.csv"
)
```
