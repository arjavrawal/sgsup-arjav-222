```python
import arcpy

# relative path to the ArcGIS Pro project
project_path = r"./Chapter 03 Files/Chapter 03.aprx"

# create a project object
project = arcpy.mp.ArcGISProject(project_path)

# access the first layout
layout = project.listLayouts()[0]
print(layout.name)
```

```python
# list all element names
for element in layout.listElements():
    print(element.name)

# retrieve the title and map frame elements
title = [e for e in layout.listElements() if e.name == "Text"][0]
map_frame = [e for e in layout.listElements() if e.name == "Map Frame"][0]

# examine the current title text
print(title.text)

# list layers in the map frame and check if they are broken
for layer in map_frame.map.listLayers():
    print(layer.name, layer.isBroken)
```

```python
# assign the counties and highways layers
counties_layer = map_frame.map.listLayers('Counties')[0]
highways_layer = map_frame.map.listLayers('Highways_Intersect')[0]

# choose a county to test
county = 'Alameda County'

# filter both layers to the selected county
counties_layer.definitionQuery = f"NAMELSAD = '{county}'"
highways_layer.definitionQuery = f"NAMELSAD = '{county}'"
```

```python
# zoom to the county extent
county_extent = map_frame.getLayerExtent(counties_layer)
map_frame.camera.setExtent(county_extent)

# set the title text
title.text = county

# save the project and export to PDF
project.save()
layout.exportToPDF(f'./{county}.pdf')
```

```python
# list of counties to export
counties = [
    'Alameda County',
    'Marin County',
    'Contra Costa County',
    'San Francisco County',
    'Santa Clara County'
]

for county in counties:
    # apply definition queries
    counties_layer.definitionQuery = f"NAMELSAD = '{county}'"
    highways_layer.definitionQuery = f"NAMELSAD = '{county}'"

    # get the county extent
    county_extent = map_frame.getLayerExtent(counties_layer)
    map_frame.camera.setExtent(county_extent)

    # update the title
    title.text = county

    # export the map to PDF using the county name
    pdf_path = f'./{county}.pdf'
    layout.exportToPDF(pdf_path)

# save changes to the project
project.save()
```
