#### Classwork 2A - List Basics

```python
# A list of coordinate tuples representing a travel route
route = [
  (35.6895, 139.6917), # Tokyo
  (34.0522, -118.2437), # Los Angeles
  (51.5074, -0.1278), # London
]
print("Travel route:", route)

# A list of elevation measurements (in meters)
elevations = [120.5, 135.2, 150.8, 168.3, 185.7, 203.1]
print("Elevation profile:", elevations)

# A list of city names
cities = ["Tokyo", "Los Angeles", "London", "Paris"]
print("Cities to visit:", cities)
```

```python
route.append((48.8566, 2.3522)) # Adding Paris coordinates
elevations.append(220.4) # Adding another elevation measurement
```

#### Classwork 2B - Dictionary Basics

```python
# Dictionary storing attributes of a city
tokyo_info = {
    "name": "Tokyo",
    "population": 13929286,
    "coordinates": (35.6895, 139.6917),
    "country": "Japan",
    "established": 1603,
}
print("Tokyo information:", tokyo_info)

# Dictionary for a geographic survey point
survey_point = {
    "point_id": "SP001",
    "latitude": 40.7589,
    "longitude": -73.9851,
    "elevation": 87.3,
    "land_cover": "urban",
    "date_surveyed": "2023-06-15",
}
print("Survey point data:", survey_point)
```

```python
survey_point.keys()
tokyo_info.values()
```

```python
print(tokyo_info["name"])
print(tokyo_info["population"])
print(tokyo_info["coordinates"])

print(survey_point.get("elevation"))
print(survey_point.get("land_cover"))
```

```python
tokyo_info["area_km2"] = 2191
tokyo_info["timezone"] = "JST"
```

```python
tokyo_info["population"] = 14000000
```

```python
tokyo_addt1 = {
    "mayor": "Yuriko Koike",
    "GDP_USD": 1650000000000
}
```

```python
tokyo_info.update(tokyo_addt1)
```

```python
print("Updated Tokyo information:", tokyo_info)
```

```python
# GPS waypoint
waypoint = {
    "id": "WP001",
    "name": "Trail Start",
    "latitude": 45.3311,
    "longitude": -121.7113,
    "elevation": 1200,
    "description": "Beginning of Pacific Crest Trail section",
    "waypoint_type": "trailhead",
    "facilities": ["parking", "restrooms", "water"],
    "difficulty": "easy",
           }
```

```python
waypoint_name = waypoint["name"]
waypoint_location = (waypoint["latitude"], waypoint["longitude"])
waypoint_elevation = waypoint["elevation"]
waypoint_facilities = waypoint["facilities"]
print(f"Waypoint: '{waypoint_name}'")
print(f"Location: {waypoint_location}")
print(f"Elevation: {waypoint_elevation} meters")
print(f"Available facilities: {waypoint_facilities}")
```
