```python
# Global variable
celebrity = "Taylor Swift"

def nfl_game():
    # Local variable
    player = "Travis Kelce"
    print(f"In the NFL game: {celebrity} is cheering for {player}!")

nfl_game()

# Trying to access local variable outside its function will result in an error.
# print(player) Error: 'player' is not defined
```

```python
# Let's bring Travis to the global spotlight
def eras_tour():
    # Travis shows up but only if Taylor brings him on stage!
    global player
    player = "Travis Kelce"
eras_tour() # Now travis is global!
print(f"{celebrity} sings 'I Can Do It With a Broken Heart'")
print(f"But look who's also on stage: {player}!")
```

```python
# parameters with default values - more flexible functions by providing fallback values when certain arguments aren't provided

def buffer_layer(layer, distance=1000, projection="WGS84"):
    return (f"Buffering {layer} with distance {distance} meters "
            f"using {projection} projection.")

# Example usage
print(buffer_layer("Roads")) # Uses defaults: 1000m, WGS84
print(buffer_layer("Rivers", distance=500)) # Overrides default distance
print(buffer_layer("Parcels", projection="NAD83")) # Overrides default projection
```

```python
# Geospatial Example: Haversine Function

# used to calculate the great-circle distance between two points on a sphere given their longitudes and latitudes

from math import radians, sin, cos, sqrt, atan2
def haversine(lat1, lon1, lat2, lon2, radius=6371.0):
    dlat = radians(lat2 - lat1)
    dlon = radians(lon2 - lon1)
    a = (
    sin(dlat / 2) ** 2
    + cos(radians(lat1)) * cos(radians(lat2)) * sin(dlon / 2) ** 2
    )
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    distance = radius * c
    return distance
# Example usage in kilometers
distance_km = haversine(35.6895, 136.6917, 34.0522, -118.2437)
print(f"Distance: {distance_km:.2f} km")
# Example usage in miles
distance_miles = haversine(35.6895, 139.6917, 34.0522, -118.2437, radius=3958.8)
print(f"Distance: {distance_miles:.2f} miles")

# nested function calls - functions can call other functions

# advanced parameter handling through *args - allows the function to accept any number of arguments and is treated as a tuple inside the function. this is particularly useful for geospatial functions that might need to handle optional metadata or attributes.

# **kwargs accepts a variable number of named parameters
```

```python
def average(*numbers):
    return sum(numbers) / len(numbers)
# Example usage
print(average(10,20,30))
print(average(5,15,25,35))
```

```python
def describe_point(latitude, longitude,**kwargs):
    description = f"Point at ({latitude}, {longitude})"
    # Add optional keyword arguments to the description
    for key, value in kwargs.items():
        description += f", {key}: {value}"
    return description
# Example usage
print(describe_point(35.6895, 139.6917, name="Tokyo", population=37400000))
```

```python
# classes are the next level of code organizing beyond functions
# classes construct objects ("blueprint" needed to create the object)
# syntax - use keyword class to create one
# attributes are class attributes - hardcoded

# create a class
class Location:
    lon = -111.9
    lat = 33.4
# create an object
tempe = Location()
print(f"Tempe is at longitude {tempe.lon} and latitude {tempe.lat}")
```

```python
# __init__ is every time an object is created. constructor sets up the object's initial state. initialization means defining attributes.
# self is the instance of the object. access and bind attributes and methods to that specific object.
# first parameter of __init__ always refers to the object being created. convention is to call it self.
# the attributes here are instance attributes because they aren't hard coded and can be changed.
# __str__ method defines how the object appears when printed

class Point:
    def __init__(self, latitude, longitude, name=None):
        self.latitude = latitude
        self.longitude = longitude
        self.name = name
    def __str__(self):
        return f"{self.name or 'Point'} ({self.latitude}, {self.longitude})"
# Example usage
point1 = Point(35.6895, 139.6917, "Tokyo")
print(point1)
```

```python
# methods are functions that are defined inside a class that operates on the class's data.

class Point:
    def __init__(self, latitude, longitude, name=None):
        self.latitude = latitude
        self.longitude = longitude
        self.name = name
    def distance_to(self, other_point):
        return haversine(
            self.latitude, self.longitude, other_point.latitude, other_point.longitude
        )
# Example usage
point1 = Point(35.6895, 139.6917, "Tokyo")
point2 = Point(34.0522, -118.2437, "Los Angeles")
# Calling the method to calculate distance
distance = point1.distance_to(point2)
print(f"Distance from {point1.name} to {point2.name}: {distance:.2f} km")
```

```python
# combine functions and classes
# create a Route class that contains a list of points and a method to calculate the total distance of the route.

class Route:
    def __init__(self, points):
        self.points = points
    def total_distance(self):
        total_dist = 0
        for i in range(len(self.points) - 1):
            total_dist += self.points[i].distance_to(self.points[i + 1])
        return total_dist
# Example route
route = Route([point1, point2])
print(f"Total distance: {route.total_distance():.2f} km")
```

```python
# Inheritance - parent class, child class. Defines a class that inherits all the methods and properties from another class

# Parent class
class Layer:
    def __init__(self, name, crs="WGS84"):
        self.name = name
        self.crs = crs
    def describe(self):
        return f"Layer: {self.name} | CRS: {self.crs}"
    
# Child class
class BaseMapLayer(Layer):
    pass

# Example
roads = BaseMapLayer("Roads", "NAD83")
print(roads.describe()) # Uses parent method
```
