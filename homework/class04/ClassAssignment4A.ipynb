```python
# create an empty list
all_values = []

# iterate through rows using a SearchCursor
for row in arcpy.da.SearchCursor(fc, ["HWY_NUM"]):
    all_values.append(row[0])

# determine the number of values
print(len(all_values))

# convert to a set to get unique values
unique_values = list(set(all_values))
print(len(unique_values))
# preview the first five unique values
print(unique_values[:5])
```

```python
clean_ints = []
for val in unique_values:
    if val and val.strip():  # ignore None or blank strings
        # remove directional letters such as 'S' or 'N'
        cleaned = val.replace("S", "").replace("N", "").strip()
        # convert to integer and append
        clean_ints.append(int(cleaned))

clean_ints = sorted(clean_ints)
print(clean_ints[:10])  # preview
```

```python
arcpy.management.AddField(
    in_table=fc,
    field_name="HWY_NUM_INT",
    field_type="SHORT",            # short integer type
)

# Debugged through Copilot - didn't use in_table the first time, oops!
```

```python
# update the new field using UpdateCursor
with arcpy.da.UpdateCursor(fc, ["HWY_NUM", "HWY_NUM_INT"]) as cursor:
    for hwy_num, hwy_int in cursor:
        if hwy_num and hwy_num.strip():
            cleaned = hwy_num.replace("S", "").replace("N", "").strip()
            new_value = int(cleaned)
        else:
            new_value = None
        # updateRow takes a tuple of the row values in the same order
        cursor.updateRow((hwy_num, new_value))
```

```python
# create an empty list
hwy_num_int = []

# iterate through rows using a SearchCursor
for row in arcpy.da.SearchCursor(fc, ["HWY_NUM_INT"]):
    hwy_num_int.append(row[0])

# determine the number of values
print(len(hwy_num_int))

# convert to a set to get unique values
unique_hwy_num_int = list(set(hwy_num_int))
print(len(unique_hwy_num_int))
# preview the first five unique values
print(unique_hwy_num_int[5:25])
```
