```python
# looping through geojson data

import json

with open('data.geojson') as f:
    data = json.load(f)
```

```python
data.keys()
```

```python
features = data['features']
```

```python
sn_count_dict = {}
```

```python
feature = features[0]
```

```python
feature['properties'].keys()
```

```python
# Store the first feature's service name and print it
service_name = feature['properties']['service_name']
service_name
```

```python
# Check if we have encountered this service before in our counter
service_name in sn_count_dict
```

```python
# Add the service_name key to the count dictionary with a count of 1
sn_count_dict[service_name] = 1
```

```python
# creating logic to increase the counter if the service_name is present OR do step 3 when encountering it for the first time

# test to see if the service name is already in there
if service_name in sn_count_dict:
    # add to the already existing value
    sn_count_dict[service_name] = sn_count_dict[service_name] + 1
else:
    # add a new key to the dictionary and populate it with a 1
    sn_count_dict[service_name] = 1
```

```python
sn_count_dict
```

```python
sn_count_dict = {}
```

```python
# reset the dictionary, just in case

sn_count_dict = {}

# iterate through each feature
for feature in features:
    # get the service name
    service_name = feature['properties']['service_name']

    # test to see if service_name is already in the dictionary
    if service_name in sn_count_dict:
        # add to the number if it's already in there
        sn_count_dict[service_name] = sn_count_dict[service_name] + 1
    else:
        # add a new key to the dictionary if not already there
        sn_count_dict[service_name] = 1
```
