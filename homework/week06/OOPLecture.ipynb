```python
# Object oriented programming is a method of structuring a program by bundling related properties and behaviors into individual objects
# Python is an OOP language -> functions and classes

# Functions are the building blocks of organized code.

# DRY approach - Don't Repeat Yourself (more efficient coding!)
# Built-in functions can be used anywhere without importation. i.e. enumerate, max, round, help.
# User-defined functions are created by the user to perform specific tasks. i.e. def my_function():

# Basic structure: the def keyword, function name, parameters in parantheses, colon, and an indented code block.
# Parameters - info that a function needs to work.
# Use the return keyword to return the output of a function. If not used, none is returned.
# Call functions by their name and arguments. Argument names do NOT always equal to parameter names.
```

```python
# Global variable
celebrity = "Taylor Swift"

def nfl_game():
    # Local variable
    player = "Travis Kelce"
    print(f"In the NFL game: {celebrity} is cheering for {player}!")

nfl_game()

# Trying to access local variable outside its function will result in an error.
print(player) # Error: 'player' is not defined
```

```python
# Let's bring Travis to the global spotlight
def eras_tour():
    # Travis shows up but only if Taylor brings him on stage!
    global player
    player = "Travis Kelce"
eras_tour() # Now travis is global!
print(f"{celebrity} sings 'I Can Do It With a Broken Heart'")
print(f"But look who's also on stage: {player}!")
```

```python
# parameters with default values - more flexible functions by providing fallback values when certain arguments aren't provided

def buffer_layer(layer, distance=1000, projection="WGS84"):
    return (f"Buffering {layer} with distance {distance} meters "
            f"using {projection} projection.")

# Example usage
print(buffer_layer("Roads")) # Uses defaults: 1000m, WGS84
print(buffer_layer("Rivers", distance=500)) # Overrides default distance
print(buffer_layer("Parcels", projection="NAD83")) # Overrides default projection
```

```python
# Geospatial Example: Haversine Function

# used to calculate the great-circle distance between two points on a sphere given their longitudes and latitudes

from math import radians, sin, cos, sqrt, atan2
def haversine(lat1, lon1, lat2, lon2, radius=6371.0):
    dlat = radians(lat2 - lat1)
    dlon = radians(lon2 - lon1)
    a = (
    sin(dlat / 2) ** 2
    + cos(radians(lat1)) * cos(radians(lat2)) * sin(dlon / 2) ** 2
    )
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    distance = radius * c
    return distance
# Example usage in kilometers
distance_km = haversine(35.6895, 136.6917, 34.0522, -118.2437)
print(f"Distance: {distance_km:.2f} km")
# Example usage in miles
distance_miles = haversine(35.6895, 139.6917, 34.0522, -118.2437, radius=3958.8)
print(f"Distance: {distance_miles:.2f} miles")

# nested function calls - functions can call other functions

# advanced parameter handling through *args - allows the function to accept any number of arguments and is treated as a tuple inside the function. this is particularly useful for geospatial functions that might need to handle optional metadata or attributes.

# **kwargs accepts a variable number of named parameters
```

```python
def average(*numbers):
    return sum(numbers) / len(numbers)
# Example usage
print(average(10,20,30))
print(average(5,15,25,35))
```

```python
def describe_point(latitude, longitude,**kwargs):
    description = f"Point at ({latitude}, {longitude})"
    # Add optional keyword arguments to the description
    for key, value in kwargs.items():
        description += f", {key}: {value}"
    return description
# Example usage
print(describe_point(35.6895, 139.6917, name="Tokyo", population=37400000))
```
