## Loops and Conditional Statements

#### For Loop

```python
# for loops - iterate through sequencies (eg. lists, tuples, strings), execute the same block of code for each item in the sequence
# assignment to a tuple is unpacking
```

```python
coordinates = [(35.6895, 139.6917), (51.5074, -0.1278), (34.0522, -118.2437)]
for lat, lon in coordinates:
    print(f"Latitude: {lat}, Longitude: {lon}")
```

```python
# processing multiple data points
def calculate_distance(lat1, lon1, lat2, lon2):
    # Distance calculation logic
    return ((lat2 - lat1)**2 + (lon2 - lon1)**2) ** 0.5
reference_point = (33.4255, -111.9400) # Example reference point (latitude, longitude) Tempe, AZ
for lat, lon in coordinates:
    distance = calculate_distance(reference_point[0], reference_point[1], lat, lon)
    print(f"Distance from reference point to ({lat}, {lon}): {distance:.2f} units")
```

```python
cities = ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix", "Kansas City"]
print("Major US Cities:")
for i, city in enumerate(cities):
    print(f"{i + 1}. {city}")
# i+1 is because enumerate starts from 0 by default (like all lists/tuples/etc do)
```

```python
# Break and Continue statements in loops
# Break statement exits a loop immediately based on external trigger

import random
best_city = cities[random.randint(0, len(cities)-1)]
for x in best_city:
    if x == 'o':
        break
    print(x)
```

```python
# Continue statements skip the current iteration and move to the next one

for x in best_city:
    if x == 'o':
        continue
    print(x)
```

```python
# Pass statement is a no-op, used as a placeholder
for x in best_city:
    if x == 'o':
        pass
    print(x)
```

```python
# range function - built-in function returns a sequence of numbers

# syntax: range([start],stop,[step]). start is optional, starting number in sequence, default is zero. stop generates number up to but not including this number. step is optional, difference between each number in sequence.

print("Some of the most popular cities are:")
for x in range(1,5,2):
    print(cities[x])
```

```python
# while loops continue executing a block of code as long as a specified condition remains true
# useful when the number of iterations required isn't known
# always make sure the condition eventually becomes false, or else this will become an infinite loop
# requires relevant variables to be ready

best_city = cities[random.randint(0, len(cities)-1)]
guess = input("Guess the best city: ")
while guess != best_city:
    print(f"You guessed {guess}.")
    print("Heh, that's mediocre at best! Try again.")
    guess = input("Wrong! Try again: ")
print (f"Correct! {guess} is the best city!")
```

```python
# another flavor of while loops

counter = 0
while counter < len(coordinates):
    lat, lon = coordinates[counter]
    print(f"Processing coordinate: ({lat},{lon})")
    counter += 1
```

```python
# Nested combined loops - place loops inside other loops. It's important to use for accessing data inside complex data structures. There is no limit to how far you can nest.

# Control statements - making decisions for your code, execute different blocks of code based on specific conditions. they help you categorize data, filter info, handle different coord systems, validate input data, respond to various geographic conditions.
```

```python
for lat, lon in coordinates:
    if lat > 0:
        print(f"{lat} is in the Northern Hemisphere.")
    elif lat < 0:
        print(f"{lat} is in the Southern Hemisphere.")
    else:
        print(f"{lat} is near the equator.")
```

```python
# Classify coordinates by quadrant
for lat, lon in coordinates:
    if lat > 0 and lon > 0:
        quadrant = "Northeast"
    elif lat > 0 and lon < 0:
        quadrant = "Northwest"
    elif lat < 0 and lon > 0:
        quadrant = "Southeast"
    else:
        quadrant = "Southwest"
    print(f"Coordinate ({lat}, {lon}) is in the {quadrant} quadrant")
```

```python
filtered_coordinates =[]
for lat, lon in coordinates:
    if lon > 0:
        filtered_coordinates.append((lat, lon))
print(f"Filtered coordinates (only with positive longitude): {filtered_coordinates}")

# This allows you to iterate through datasets while making intelligent decisions about each piece of data you encounter.

# Counting and Analyzing Data
southern_count = 0
for lat, lon in coordinates:
    if lat < 0:
        southern_count += 1
print(f"Number of coordinates in the Southern Hemisphere: {southern_count}")
```

```python
# Building summary statistics

northern_count = 0
for lat, lon in coordinates:
    if lat > 0:
        northern_count += 1
print(f"Number of coordinates in the Northern Hemisphere: {northern_count}")

eastern_count = 0
for lat, lon in coordinates:
    if lon > 0:
        eastern_count += 1
print(f"Eastern Count: {eastern_count}")

western_count = 0
for lat, lon in coordinates:
    if lon < 0:
        western_count += 1
print(f"Western Count: {western_count}")
```

```python
# Summary Statistics

analysis_coordinates = [
    (40.7128, -74.0060) # New York
    (-33.8688, 151.2093) # Sydney
    (51.5074, -0.1278), # London
    (-1.2921, 36.8219) # Nairobi
    (35.6762, 139.6503) # Tokyo
]

# Initialize counters
northern_count = 0
southern_count = 0
eastern_count = 0
western_count = 0
valid_coordinates = []
print("coordinate analysis:")
print("-" * 40)
for lat, lon in analysis_coordinates:
    # Validate Coordinates (basic check)
    if -90 <= lat <= and -180 <= lon <= 180:
        valid_coordinates.append((lat, lon))
        # Count by hemisphere
        if lat >= 0:
            northern_count += 1
        else:
            southern_count += 1
        # Count by longitude
        if lon >= 0:
            eastern_count += 1
        else:
            western_count += 1
print(f"Valid: ({lat:7.4f}, {lon:8.4f})")
else:
    print(f"Invalid: ({lat}), ({lon})- coordinates out of range")
print (f"\nSummary:")
print(f"Valid Coordinates:")
```

```python
# Switch statements can be used to select one of many code blocks to be executed.

def get_region_by_country(country_name):
    match country_name:
        case "Japan" | "Nippon":
            return "East Asia"
        case "United States" | "USA" | "Canada" | "Mexico":
            return "North America"
        case "United Kingdom" | "UK":
            return "Europe"
        case "Australia":
            return "Oceania"
        case "Kenya":
            return "Africa"
        case "India" | "Bharat":
            return "Southeast Asia"
        case _:
            return "Unknown Region"
        
country = input("Enter a country name: ")
print(get_region_by_country(country))
```
